```python
import numpy as np

# Function to display the matrix
def print_matrix(M, title):
    print(f"\n{title}")
    print("-" * 30)

    for row in M:
        print("[ " + "  ".join(f"{x:8.2f}" for x in row) + " ]")

def print_list(l, title):
  print(f"\n{title}")
  print("-" * 30)
  print("[ " + "  ".join(f"{x:8.2f}" for x in l) + " ]")
```

```python
# Gauss Elimination with Partial Pivoting -- Umanga Khatri
def gauss_elimination_pp(A, b):

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    rows, cols = A.shape

    if rows != cols:
        raise ValueError("Matrix A must be square.")

    if len(b) != rows:
        raise ValueError("Size of b must match the number of rows in A.")

    n = rows

    M = np.column_stack((A, b))

    print_matrix(M, "Initial Augmented Matrix [A | b]")

    for i in range(n - 1):

        max_row = i + np.argmax(np.abs(M[i:, i]))

        if np.isclose(M[max_row, i], 0):
            raise ValueError(
                "The matrix is singular or has no unique solution."
            )

        if max_row != i:

            M[[i, max_row]] = M[[max_row, i]]

            print_matrix(
                M,
                f"Step {i + 1}: R{i + 1} <-> R{max_row + 1}"
            )

        for j in range(i + 1, n):

            factor = M[j, i] / M[i, i]

            M[j] = M[j] - factor * M[i]

            print_matrix(
                M,
                f"R{j + 1} = R{j + 1} - ({factor:.4f})R{i + 1}"
            )

    if np.isclose(M[n - 1, n - 1], 0):
        raise ValueError(
            "The matrix is singular or has no unique solution."
        )

    x = np.zeros(n)

    print("\nBack Substitution")
    print("-" * 60)

    for i in range(n - 1, -1, -1):

        x[i] = (
            M[i, -1]
            - np.dot(M[i, i + 1:n], x[i + 1:n])
        ) / M[i, i]

        print(f"x{i + 1} = {x[i]:.6f}")

    return x


A = np.array([
    [2, -6, -1],
    [-3, -1, 7],
    [-8, 1, -2]
], dtype=float)

b = np.array([-38, -34, -20], dtype=float)


try:

    solution = gauss_elimination_pp(A, b)

    print("\nFinal Solution")
    print("-" * 60)

    for i, value in enumerate(solution, start=1):
        print(f"x{i} = {value:.6f}")

except ValueError as error:
    print("\nError:", error)
```

```python
# Gauss-Jordan Elimination -- Yunesh Timsina
def gauss_jordan(A, b):

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    n = len(b)

    if A.shape != (n, n):
        raise ValueError("A must be a square matrix.")

    M = np.column_stack((A, b))

    print_matrix(M, "Initial Augmented Matrix")

    for i in range(n):

        max_row = i + np.argmax(np.abs(M[i:, i]))

        if np.isclose(M[max_row, i], 0):
            raise ValueError(
                "The matrix is singular or has no unique solution."
            )

        if max_row != i:

            M[[i, max_row]] = M[[max_row, i]]

            print_matrix(
                M,
                f"Step {i + 1}: R{i + 1} <-> R{max_row + 1}"
            )

        pivot = M[i, i]

        M[i] = M[i] / pivot

        print_matrix(
            M,
            f"Step {i + 1}: R{i + 1} = R{i + 1} / ({pivot:.2f})"
        )

        for j in range(n):

            if i != j:

                factor = M[j, i]

                if not np.isclose(factor, 0):

                    M[j] = M[j] - factor * M[i]

                    print_matrix(
                        M,
                        f"R{j + 1} = R{j + 1} - ({factor:.2f})R{i + 1}"
                    )

    return M[:, -1]

A = np.array([
    [2, 3, -1],
    [1, 1, 1],
    [-2, 5, 2]
], dtype=float)

b = np.array([54, 6, -3], dtype=float)

solution = gauss_jordan(A, b)

print("\nFinal Solution")
print("-" * 45)

for i, value in enumerate(solution, start=1):
    print(f"x{i} = {value:.2f}")
```

```python
def swap_rows(A, b):
    A = A.copy()
    b = b.copy()

    n = len(b)

    for i in range(n):
        max_row = i

        for j in range(i + 1, n):
            if abs(A[j][i]) > abs(A[max_row][i]):
                max_row = j

        if max_row != i:
            A[[i, max_row]] = A[[max_row, i]]
            b[[i, max_row]] = b[[max_row, i]]

    return A, b
```

```python
# Gauss Seidel Method -- Yunesh Timsina
def gauss_seidel(A, b, x0, tol=0.0005, max_iter=10):

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    x = np.array(x0, dtype=float)

    n = len(b)

    if A.shape != (n, n):
        raise ValueError("A must be a square matrix.")

    if len(x) != n:
        raise ValueError("Initial guess must have the same size as b.")

    if np.any(np.isclose(np.diag(A), 0)):
        raise ValueError("Diagonal elements cannot be zero.")

    A, b = swap_rows(A, b)

    print_matrix(A, "\nSwapped Matrix")
    print_list(b, "\nSwapped RHS")

    iterations = []

    for k in range(max_iter):

        x_old = np.copy(x)

        for i in range(n):
            sum1 = np.dot(A[i, :i], x[:i])
            sum2 = np.dot(A[i, i + 1:], x_old[i + 1:])

            x[i] = (b[i] - sum1 - sum2) / A[i, i]

        errors = np.zeros(n)

        for i in range(n):
            if x[i] != 0:
                errors[i] = abs((x[i] - x_old[i]) / x[i]) * 100

        iterations.append([k + 1, *x.copy(), *errors])

        if np.max(errors) < tol:
            break

    width = 13

    print("\nGauss-Seidel Iterations")
    print("=" * (width + 2 * n * width))

    print(f"{'Iteration':^13}", end="")

    for i in range(n):
        print(f"{'x' + str(i + 1):^13}", end="")

    for i in range(n):
        print(f"{'E_x' + str(i + 1):^13}", end="")

    print()

    print("-" * (width + 2 * n * width))

    for row in iterations:

        print(f"{int(row[0]):^13}", end="")

        for value in row[1:n + 1]:
            print(f"{value:^13.6f}", end="")

        for value in row[n + 1:]:
            print(f"{value:^13.6f}", end="")

        print()

    print("=" * (width + 2 * n * width))

    if np.max(errors) < tol:
        print(f"Converged after {k + 1} iterations")
    else:
        print("Maximum iterations reached")

    return x


A = np.array([
    [-3, 1, 15],
    [6, -2, 1],
    [5, 10, 1]
], dtype=float)

b = np.array([44, 5, 28], dtype=float)

x0 = np.zeros(len(b))

solution = gauss_seidel(A, b, x0)

print("\nFinal Solution")
print("=" * 30)

for i, value in enumerate(solution, start=1):
    print(f"x{i} = {value:.6f}")
```
